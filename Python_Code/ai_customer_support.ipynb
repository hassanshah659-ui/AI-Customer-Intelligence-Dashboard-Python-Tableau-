KPI's

```python
# 1. Resolution Rate
resolution_rate = df['Is_Resolved'].mean() * 100

# Filter for resolved tickets for time-based KPIs
resolved_tickets = df[df['Time to Resolution'].notna()].copy()

# 2. Average Resolution Time
# Assuming HH:MM:SS part of 'Time to Resolution' represents the duration in HH:MM:SS format.
if not resolved_tickets.empty:
    resolved_tickets['Resolution_Duration'] = pd.to_timedelta(resolved_tickets['Time to Resolution'].str.split(' ').str[1])
    avg_resolution_time = resolved_tickets['Resolution_Duration'].mean().total_seconds() / 3600
else:
    avg_resolution_time = 0.0

# 3. Average Response Time
# Filter for tickets that received a response
responded_tickets = df[df['First Response Time'] != 'No Response'].copy()
# Assuming HH:MM:SS part of 'First Response Time' represents the duration in HH:MM:SS format.
if not responded_tickets.empty:
    responded_tickets['Response_Duration'] = pd.to_timedelta(responded_tickets['First Response Time'].str.split(' ').str[1])
    avg_response_time = responded_tickets['Response_Duration'].mean().total_seconds() / 3600
else:
    avg_response_time = 0.0

# 4. Average CSAT
avg_csat = df['Customer Satisfaction Rating'].mean()

# 5. SLA Compliance Rate
# Define hypothetical SLA targets in hours based on Ticket Priority
sla_targets = {
    'Critical': 4,   # E.g., Critical tickets should be resolved within 4 hours
    'High': 8,     # E.g., High tickets should be resolved within 8 hours
    'Medium': 24,  # E.g., Medium tickets should be resolved within 24 hours
    'Low': 48      # E.g., Low tickets should be resolved within 48 hours
}

if not resolved_tickets.empty:
    resolved_tickets['SLA_Met'] = resolved_tickets.apply(
        lambda row: 1 if row['Ticket Priority'] in sla_targets and \
                       row['Resolution_Duration'].total_seconds() / 3600 <= sla_targets[row['Ticket Priority']] else 0,
        axis=1
    )
    sla_rate = (resolved_tickets['SLA_Met'].sum() / len(resolved_tickets)) * 100
else:
    sla_rate = 0.0

print(" ----------- KPI SUMMARY ----------- ")

print(f"Resolution Rate: {resolution_rate:.2f}%")

print(f"Average Resolution Time: {avg_resolution_time:.2f} hours")

print(f"Average Response Time: {avg_response_time:.2f} hours")

print(f"Average CSAT: {avg_csat:.2f}")

print(f"SLA Compliance Rate: {sla_rate:.2f}%")
```
